Cell-1 Install, import and download dataset

```python
!pip install -q kagglehub

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import kagglehub
import os


path = kagglehub.dataset_download(
    "manishabhatt22/marketing-campaign-performance-dataset"
)


csv_files = []

for root, folders, files in os.walk(path):
    for file in files:
        if file.endswith(".csv"):
            csv_files.append(os.path.join(root, file))

if len(csv_files) == 0:
    raise Exception("CSV file not found")

df = pd.read_csv(csv_files[0])

print("Dataset downloaded successfully")
print("Rows and columns:", df.shape)

print(df.head())
```

Cell-2 Data cleaning and standardisation

```python

df.columns = df.columns.str.strip().str.lower()


df = df.drop_duplicates()


df["channel_used"] = df["channel_used"].astype(str).str.strip().str.title()


df["campaign_type"] = df["campaign_type"].astype(str).str.strip().str.title()


df["target_audience"] = df["target_audience"].astype(str).str.strip().str.title()


df["campaign_name"] = (
    df["campaign_type"] + "_" + df["channel_used"]
)


df["date"] = pd.to_datetime(df["date"], errors="coerce")


df["acquisition_cost"] = pd.to_numeric(
    df["acquisition_cost"].astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False),
    errors="coerce"
)


for col in ["clicks", "impressions", "conversion_rate", "roi"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")


print("Missing values:")
print(df.isnull().sum())


df = df.dropna(subset=[
    "date", "channel_used", "campaign_type",
    "target_audience", "clicks", "impressions",
    "acquisition_cost", "conversion_rate", "roi"
])

df = df[
    (df["impressions"] > 0) &
    (df["clicks"] >= 0) &
    (df["clicks"] <= df["impressions"]) &
    (df["acquisition_cost"] > 0) &
    (df["conversion_rate"] >= 0)
].copy()

print("Cleaned dataset:")
print(df.head())
print("Final dataset shape:", df.shape)
```

Cell-3 Validate campaign grain and reconcile metrics

```python

print("Total records:", len(df))

print("Unique campaigns:", df["campaign_id"].nunique())

print("Unique channels:", df["channel_used"].nunique())

print("Duplicate records:", df.duplicated().sum())


total_spend = df["acquisition_cost"].sum()

total_clicks = df["clicks"].sum()

total_impressions = df["impressions"].sum()

print("\nCAMPAIGN RECONCILIATION")

print("Total Spend:", total_spend)

print("Total Clicks:", total_clicks)

print("Total Impressions:", total_impressions)


print("\nMETRICS NOT AVAILABLE IN ORIGINAL DATASET")

print("Actual Sessions: Not available")

print("Actual Leads: Not available")

print("Actual Orders: Not available")

print("Actual Customer Touchpoints: Not available")

print("Conversions will be estimated in the next cell.")
```

Cell - 4 Calculate all campaign KPIs

```python

if df["conversion_rate"].median() > 1:
    df["conversion_rate"] = df["conversion_rate"] / 100


if df["roi"].abs().median() > 1:
    df["roi"] = df["roi"] / 100


df["conversions_estimated"] = (
    df["clicks"] * df["conversion_rate"]
)


df["roas"] = 1 + df["roi"]


df["revenue_estimated"] = (
    df["acquisition_cost"] * df["roas"]
)


df["ctr"] = (
    df["clicks"] / df["impressions"]
) * 100


df["cpc"] = (
    df["acquisition_cost"] /
    df["clicks"].replace(0, np.nan)
)


df["cvr"] = (
    df["conversions_estimated"] /
    df["clicks"].replace(0, np.nan)
) * 100


df["cpa"] = (
    df["acquisition_cost"] /
    df["conversions_estimated"].replace(0, np.nan)
)


df["contribution"] = (
    df["revenue_estimated"] /
    df["revenue_estimated"].sum()
) * 100


kpi_model = df[[
    "campaign_name",
    "channel_used",
    "impressions",
    "clicks",
    "acquisition_cost",
    "conversions_estimated",
    "ctr",
    "cpc",
    "cvr",
    "cpa",
    "roas",
    "contribution"
]]

print(kpi_model.head(10).round(2))

print("\nTOTAL CAMPAIGN PERFORMANCE")

print("Spend:", round(total_spend, 2))

print("Estimated Conversions:",
      round(df["conversions_estimated"].sum(), 2))

print("Estimated Revenue:",
      round(df["revenue_estimated"].sum(), 2))

print("Overall ROAS:",
      round(df["revenue_estimated"].sum() / total_spend, 2))
```

Cell-5 Channel performance summary

```python
channel_summary = df.groupby("channel_used").agg({
    "impressions": "sum",
    "clicks": "sum",
    "acquisition_cost": "sum",
    "conversions_estimated": "sum",
    "revenue_estimated": "sum"
})


channel_summary["CTR"] = (
    channel_summary["clicks"] /
    channel_summary["impressions"] * 100
)

channel_summary["CPC"] = (
    channel_summary["acquisition_cost"] /
    channel_summary["clicks"]
)

channel_summary["CVR"] = (
    channel_summary["conversions_estimated"] /
    channel_summary["clicks"] * 100
)

channel_summary["CPA"] = (
    channel_summary["acquisition_cost"] /
    channel_summary["conversions_estimated"]
)

channel_summary["ROAS"] = (
    channel_summary["revenue_estimated"] /
    channel_summary["acquisition_cost"]
)

channel_summary["Contribution"] = (
    channel_summary["revenue_estimated"] /
    channel_summary["revenue_estimated"].sum() * 100
)

print("CHANNEL PERFORMANCE")
print(channel_summary.round(2))
```

Cell-6 NumPy uncertainty and campaign distributions

```python

np.random.seed(42)


for channel in df["channel_used"].unique():

    values = df[
        df["channel_used"] == channel
    ]["roas"].dropna().values

    if len(values) == 0:
        continue


    samples = np.random.choice(
        values,
        size=(1000, len(values)),
        replace=True
    )

    averages = np.mean(samples, axis=1)

    lower = np.percentile(averages, 2.5)

    upper = np.percentile(averages, 97.5)

    print("\nChannel:", channel)

    print("Mean ROAS:", round(np.mean(values), 2))

    print("Standard Deviation:", round(np.std(values), 2))

    print("95% Bootstrap Interval:",
          round(lower, 2), "to", round(upper, 2))



plt.figure(figsize=(10, 5))

channel_data = []

channel_names = []

for channel in df["channel_used"].unique():

    values = df[
        df["channel_used"] == channel
    ]["roas"].dropna().values

    if len(values) > 0:
        channel_data.append(values)
        channel_names.append(channel)

plt.boxplot(channel_data, tick_labels=channel_names)

plt.title("ROAS Distribution Across Channels")

plt.xlabel("Marketing Channels")

plt.ylabel("ROAS")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

Cell-7

```python

df["month"] = df["date"].dt.to_period("M").astype(str)


time_summary = df.groupby("month").agg({
    "impressions": "sum",
    "clicks": "sum",
    "acquisition_cost": "sum",
    "conversions_estimated": "sum",
    "revenue_estimated": "sum"
})

time_summary["ROAS"] = (
    time_summary["revenue_estimated"] /
    time_summary["acquisition_cost"]
)

time_summary["CTR"] = (
    time_summary["clicks"] /
    time_summary["impressions"] * 100
)

print("TIME PERFORMANCE SUMMARY")

print(time_summary.round(2))
```

Cell-8

```python
audience_summary = df.groupby("target_audience").agg({
    "impressions": "sum",
    "clicks": "sum",
    "acquisition_cost": "sum",
    "conversions_estimated": "sum",
    "revenue_estimated": "sum"
})

audience_summary["CTR"] = (
    audience_summary["clicks"] /
    audience_summary["impressions"] * 100
)

audience_summary["CPA"] = (
    audience_summary["acquisition_cost"] /
    audience_summary["conversions_estimated"]
)

audience_summary["ROAS"] = (
    audience_summary["revenue_estimated"] /
    audience_summary["acquisition_cost"]
)

print("AUDIENCE PERFORMANCE SUMMARY")

print(audience_summary.round(2))
```

Cell-9

```python

np.random.seed(42)

channels = df["channel_used"].unique()

touchpoints = pd.DataFrame()

touchpoints["campaign_id"] = df["campaign_id"].values

touchpoints["revenue"] = df["revenue_estimated"].values


touchpoints["first_channel"] = np.random.choice(
    channels,
    size=len(df)
)


touchpoints["middle_channel"] = np.random.choice(
    channels,
    size=len(df)
)


touchpoints["last_channel"] = df["channel_used"].values


last_touch = touchpoints.groupby(
    "last_channel"
)["revenue"].sum()


first = touchpoints[[
    "first_channel", "revenue"
]].rename(columns={
    "first_channel": "channel"
})

middle = touchpoints[[
    "middle_channel", "revenue"
]].rename(columns={
    "middle_channel": "channel"
})

last = touchpoints[[
    "last_channel", "revenue"
]].rename(columns={
    "last_channel": "channel"
})

all_touches = pd.concat([first, middle, last])

all_touches["credit"] = all_touches["revenue"] / 3

linear_touch = all_touches.groupby(
    "channel"
)["credit"].sum()


attribution = pd.DataFrame({
    "Last Touch": last_touch,
    "Linear Attribution": linear_touch
}).fillna(0)

print("ATTRIBUTION COMPARISON")

print(attribution.round(2))


attribution.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Last Touch vs Linear Attribution")

plt.xlabel("Marketing Channel")

plt.ylabel("Attributed Revenue")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

Cell-10

```python

budget = channel_summary["acquisition_cost"].copy()

total_budget = budget.sum()


best_channel = channel_summary["ROAS"].idxmax()

worst_channel = channel_summary["ROAS"].idxmin()

new_budget = budget.copy()

if best_channel != worst_channel:

    transfer = min(
        total_budget * 0.10,

        max(0, budget[worst_channel] - total_budget * 0.05),

        max(0, total_budget * 0.40 - budget[best_channel])
    )

    new_budget[worst_channel] -= transfer

    new_budget[best_channel] += transfer

else:
    transfer = 0


budget_comparison = pd.DataFrame({
    "Current Budget": budget,
    "New Budget": new_budget
})

print("BUDGET REALLOCATION")

print(budget_comparison.round(2))

print("\nBest ROAS Channel:", best_channel)

print("Lowest ROAS Channel:", worst_channel)

print("Budget Transferred:", round(transfer, 2))

print("Original Total Budget:", round(budget.sum(), 2))

print("New Total Budget:", round(new_budget.sum(), 2))

print("Budget Difference:",
      round(new_budget.sum() - budget.sum(), 2))
```
